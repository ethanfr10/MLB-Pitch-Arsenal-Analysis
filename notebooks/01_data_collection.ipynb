# 01 - Statcast Data Collection

This notebook collects an initial sample of MLB Statcast data and explores its structure before building the full dataset.

```python
import pandas as pd
import numpy as np
from pybaseball import statcast

pd.set_option("display.max_columns", None)
```

## Initial Data Sample

Start with a small date range to understand the Statcast data before collecting full seasons.

```python
sample = statcast(
    start_dt="2025-06-01",
    end_dt="2025-06-03"
)
```

## Relevant Variables

Keep an initial set of variables related to pitch characteristics, movement, location, context, and outcomes.

```python
columns_to_keep = [
    "game_date",
    "player_name",
    "pitcher",
    "batter",
    "pitch_type",
    "release_speed",
    "release_spin_rate",
    "pfx_x",
    "pfx_z",
    "release_pos_x",
    "release_pos_z",
    "release_extension",
    "plate_x",
    "plate_z",
    "zone",
    "stand",
    "p_throws",
    "balls",
    "strikes",
    "description",
    "events"
]
```

```python
pitch_data = sample[columns_to_keep].copy()
pitch_data.head(10)
```

```python
print("Number of pitches:", len(pitch_data))
print("Number of columns:", len(pitch_data.columns))
pitch_data.info()
```

```python
pitcher_name = "Crochet, Garrett"
one_pitcher = pitch_data[
    pitch_data["player_name"] == pitcher_name
].copy()
one_pitcher.head()
```

```python
one_pitcher["pitch_type"].value_counts()
```

## Arsenal Summary

Summarize Crochet's pitch usage and average characteristics by pitch type.

```python
arsenal = one_pitcher.groupby("pitch_type").agg(
    pitches=("pitch_type", "count"),
    avg_velocity=("release_speed", "mean"),
    avg_spin_rate=("release_spin_rate", "mean"),
    avg_horizontal_movement=("pfx_x", "mean"),
    avg_vertical_movement=("pfx_z", "mean")
)
arsenal.round(2)
```

```python
arsenal["usage_pct"] = (
    arsenal["pitches"] / arsenal["pitches"].sum()
) * 100
arsenal = arsenal.sort_values(
    by="pitches",
    ascending=False
)
arsenal.round(2)
```

## Saving the Sample

The smaller Statcast dataset will be saved locally so that it can be reused without downloading the same data again.

Parquet format is being used because it is efficient for storing larger datasets and preserves data types better than CSV.

```python
pitch_data.to_parquet(
    "../data/raw/statcast_sample.parquet",
    index=False
)
```

## Initial Takeaways

- Statcast provides pitch level information covering pitch characteristics, movement, location, game context, and outcomes.
- The dataset contains considerably more variables than will be necessary for the initial analysis.
- `description` represents the result of an individual pitch, while `events` generally represents the final result of the plate appearance.
- Missing values do not always represent data quality problems. Some variables, such as `events`, are only expected to contain values for certain pitches.
- Pitch type will need to be considered when comparing pitch characteristics because different pitch types are designed to behave differently.
- A small sample is sufficient for understanding the dataset, but a much larger historical dataset will eventually be required for modeling.
