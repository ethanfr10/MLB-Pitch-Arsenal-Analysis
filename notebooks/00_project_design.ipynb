# MLB Pitch Quality & Arsenal Analysis

## Project Overview

The goal of this project is to develop an MLB wide framework for evaluating individual pitch quality and pitcher arsenals using Statcast data.

Rather than evaluating pitches only by their observed results, the project will examine the underlying characteristics that contribute to pitch effectiveness, including velocity, movement, spin, release characteristics, location, and game context.

The project will eventually expand from individual pitch evaluation to pitch similarity, arsenal construction, and pitcher level analytical scouting reports.

## Research Questions

This project will investigate several related questions:

1. Which physical characteristics are most associated with pitch effectiveness?

2. Can expected pitch effectiveness be estimated using pitch characteristics rather than relying solely on observed outcomes?

3. How does pitch quality vary across pitch types and pitchers?

4. Which MLB pitches are most similar based on velocity, movement, spin, and release characteristics?

5. How do pitches complement one another within a pitcher's arsenal?

6. Can pitcher arsenals be evaluated based on pitch quality, pitch separation, and pitch diversity?

7. Can the resulting models identify strengths, weaknesses, and potential areas for pitcher development?

## Unit of Analysis

The primary unit of analysis will be the individual pitch.

Each observation in the primary dataset will represent one pitch thrown in an MLB game.

Secondary levels of analysis will include:

- Individual pitch
- Pitcher pitch type combination
- Pitcher arsenal
- Pitcher season

Pitch level modeling will provide the foundation for higher level pitcher and arsenal evaluations.

## Modeling Framework
Pitch effectiveness can be evaluated from multiple perspectives.

### Stuff
Evaluates the physical characteristics of a pitch, such as velocity, movement, spin, and release characteristics.

### Command
Evaluates where a pitch is located and how effectively the pitcher places the pitch.

### Overall Pitch Quality
Combines pitch characteristics, location, and contextual information to estimate expected pitch effectiveness.

These components should initially be modeled separately so that physical pitch quality is not confused with location or game context.

## Target Variable — To Be Determined

The target variable for the initial pitch quality model has not yet been selected.

Potential approaches include:

- Whiff probability
- Called strike probability
- Expected run value
- Expected pitch outcome
- A multi component pitch effectiveness metric

The final target will be selected after exploring the available Statcast variables and determining which outcome best represents the baseball question being modeled.

## Planned Feature Groups

Features will initially be organized into four groups.

### 1. Raw Pitch Characteristics
Direct measurements describing the pitch.

Examples:
- Velocity
- Spin rate
- Horizontal movement
- Vertical movement
- Release height
- Release side
- Extension
- Spin axis

### 2. Location Features
Variables describing where the pitch crosses the plate.

Examples:
- Plate X
- Plate Z
- Distance from the center of the strike zone
- Relative vertical location within the batter's strike zone

### 3. Contextual Features
Variables describing the situation in which the pitch was thrown.

Examples:
- Batter handedness
- Pitcher handedness
- Count
- Platoon matchup
- Base/out state

### 4. Engineered Arsenal Features
Variables created by comparing a pitch with the pitcher's other offerings.

Potential examples:
- Velocity separation
- Horizontal movement separation
- Vertical movement separation
- Release point similarity
- Pitch usage
- Primary fastball characteristics
- Difference from primary fastball

These engineered features will be developed after the raw Statcast data has been explored.

## Planned Modeling Architecture

The project will be developed as a series of connected analytical components.

### Model 1 — Pitch Quality
Estimate expected pitch effectiveness using physical pitch characteristics.

### Model 2 — Command / Location
Evaluate the relationship between pitch location and expected effectiveness.

### Model 3 — Overall Pitch Effectiveness
Combine pitch characteristics, location, and relevant context to estimate overall expected pitch effectiveness.

### Model 4 — Pitch Similarity
Identify comparable pitches across MLB using standardized pitch characteristics and similarity methods.

### Model 5 — Arsenal Analysis
Evaluate how a pitcher's offerings work together based on pitch quality, velocity separation, movement separation, release characteristics, usage, and pitch diversity.

The outputs of these components will eventually be aggregated into pitcher level analytical profiles.

## Validation Strategy

Model evaluation will prioritize out of sample performance and baseball interpretability.

Potential evaluation methods include:

### Predictive Performance
Depending on the selected target:
- Log Loss
- ROC-AUC
- Brier Score
- Calibration
- Mean Absolute Error
- Root Mean Squared Error

### Baseball Validation
Model outputs should also demonstrate meaningful relationships with future or independent performance.

Potential tests include:
- Year to year stability
- Relationship with future whiff rate
- Relationship with future run value
- Performance by pitch type
- Performance by pitcher handedness
- Performance against opposite and same handed hitters

### Train/Test Design
Where possible, chronological train/test splits will be preferred over random splits to better simulate real world evaluation of future pitches.

Model quality will not be judged solely by a single predictive metric.

## Intended Final Product

The completed project should provide an MLB wide pitch and arsenal evaluation system rather than an analysis of a single pitcher.

The final system should allow a user to:

1. Search for an MLB pitcher.
2. View the pitcher's arsenal and pitch usage.
3. Evaluate the quality of each pitch.
4. Compare each pitch with MLB averages.
5. Find similar pitches thrown by other pitchers.
6. Examine velocity and movement separation within the arsenal.
7. Identify arsenal strengths and potential weaknesses.
8. Generate an analytical pitcher profile.

Individual pitchers will be used as case studies to demonstrate the system rather than serving as the sole focus of the analysis.
